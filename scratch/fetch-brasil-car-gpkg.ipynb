Feature count printed after each chunk download.

Validation after merging to check for duplicate geometries (using WKT) or duplicate feature IDs (if present).

```python
import os
import subprocess
import time
import xml.etree.ElementTree as ET
from datetime import datetime
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed
import geopandas as gpd

# Suppress certificate validation warnings
os.environ["CURL_CA_BUNDLE"] = ""

# Setup output directory
TODAY = datetime.today().strftime("%Y%m%d")
BASE_DIR = Path(f"CAR_data_pull/{TODAY}")
BASE_DIR.mkdir(parents=True, exist_ok=True)

def download_chunk(uf_code, chunk_size, i, max_retries=5):
    layer = f"sicar:sicar_imoveis_{uf_code}"
    output_dir = BASE_DIR / uf_code
    output_dir.mkdir(exist_ok=True)

    start = i * chunk_size
    url = (
        "https://geoserver.car.gov.br/geoserver/sicar/ows?"
        f"service=WFS&version=2.0.0&request=GetFeature&typeName={layer}"
        f"&outputFormat=application/gml%2Bxml%3B%20version%3D3.2"
        f"&count={chunk_size}&startIndex={start}"
    )
    file_path = output_dir / f"{uf_code}_chunk_{i}.gml"
    print(f"Yo! Chunk {i}: {file_path.name}")

    for attempt in range(1, max_retries + 1):
        subprocess.run([
            "curl", "--insecure", "-s", "-L", "-A", "Mozilla/5.0",
            "-o", str(file_path), url
        ], stdout=subprocess.PIPE, stderr=subprocess.PIPE)

        if file_path.exists() and file_path.stat().st_size > 1000:
            try:
                tree = ET.parse(file_path)
                root = tree.getroot()
                ns = {"wfs": "http://www.opengis.net/wfs/2.0"}
                features = root.findall(".//wfs:member", ns)
                if len(features) == 0:
                    print(f"Chunk {i} appears empty — stopping.")
                    file_path.unlink(missing_ok=True)
                    return (i, False)
                print(f"Chunk {i} has {len(features)} features.")
                return (i, True)
            except Exception as e:
                print(f"!!! XML parse error in chunk {i}: {e}")
        else:
            print(f"Chunk {i} attempt {attempt} failed")
            if file_path.exists():
                with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
                    print(f"Partial content: {f.read(500)[:250]}...")
                file_path.unlink()
            time.sleep(5)

    print(f"!! Chunk {i} failed after {max_retries} retries.")
    return (i, False)

def download_fixed_chunks_batched(uf_code, chunk_size=1000, max_chunks=5000, batch_size=10):
    for batch_start in range(0, max_chunks, batch_size):
        print(f"\nDownloading batch starting at chunk {batch_start}")
        futures = []
        with ThreadPoolExecutor(max_workers=batch_size) as executor:
            for i in range(batch_start, batch_start + batch_size):
                futures.append(executor.submit(download_chunk, uf_code, chunk_size, i))
            for future in as_completed(futures):
                index, success = future.result()
                if not success:
                    print(f"Stopping at chunk {index}")
                    return

def convert_chunks_to_gpkg(uf_code):
    input_dir = BASE_DIR / uf_code
    chunks = sorted(input_dir.glob("*.gml"))
    output_path = BASE_DIR / f"sicar_imoveis_{uf_code}.gpkg"

    if not chunks:
        print(f"No GML files for {uf_code}")
        return

    print(f"Converting {len(chunks)} chunks to {output_path}")
    for i, gml in enumerate(chunks):
        if i == 0:
            os.system(f"ogr2ogr -f GPKG {output_path} {gml}")
        else:
            os.system(f"ogr2ogr -f GPKG -update -append {output_path} {gml} -nln sicar_imoveis_{uf_code}")

    size = output_path.stat().st_size / (1024 * 1024)
    print(f"YES Final GPKG: {output_path} ({size:.2f} MB)")

    print(f"Validating for duplicate geometries and IDs...")
    try:
        gdf = gpd.read_file(output_path)
        dup_geoms = gdf.duplicated(subset="geometry")
        print(f"Exact duplicate geometries: {dup_geoms.sum()}")
        if 'id' in gdf.columns:
            dup_ids = gdf.duplicated(subset='id')
            print(f"Duplicate IDs: {dup_ids.sum()}")
        else:
            print("ℹNo 'id' field for ID check.")
    except Exception as e:
        print(f"!!!!! GPKG validation failed: {e}")

def process_state(uf_code):
    print(f"\n Processing {uf_code.upper()}")
    download_fixed_chunks_batched(uf_code)
    convert_chunks_to_gpkg(uf_code)

# To run:
all_uf_codes = ["al", "am", "ap", "ba", "ce", "df", "es", "go", "ma", "mt", "ms",
                 "pa", "pb", "pr", "pe", "pi", "rj", "rn", "ro", "rr", "sc", "se", "sp", "to",
                 "mg", "rs", "ac"]
for uf in all_uf_codes:
    process_state(uf)
```
