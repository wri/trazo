```python
import os
import random
import shutil
import re

# Set random seed for reproducibility
random.seed(42)

# Root directories
base_paths = {
    "Mato Grosso Extended Soy": r"G:\My Drive\Active Learning Pipeline - Annotations\Sampling Experiment\Mato Grosso Extended Soy"#,
    #"Chiquitano Long Strip Soy": r"G:\My Drive\Active Learning Pipeline - Annotations\Sampling Experiment\Chiquitano Long Strip Soy"
}

# Mapping between base folders and their 100-chip export folders
tif_sources = {
    "Mato Grosso Extended Soy": r"G:\My Drive\Active Learning Pipeline - Annotations\Sampling Experiment\Mato Grosso Extended Soy\geeexports_asuwriregions_MGER2random100from1000"#,
    #"Chiquitano Long Strip Soy": r"G:\My Drive\Active Learning Pipeline - Annotations\Sampling Experiment\Chiquitano Long Strip Soy\geeexports_asuwriregions_CLSR1random100from1000"
}

def move_chips(source_folder, dest_folder, num_chips_to_move):
    # List all .tif files in the source
    tifs = [f for f in os.listdir(source_folder) if f.lower().endswith(".tif")]
    if num_chips_to_move > len(tifs):
        raise ValueError(f"Not enough TIFFs to move from {source_folder}. Requested {num_chips_to_move}, found {len(tifs)}")

    # Randomly select without replacement
    selected = random.sample(tifs, num_chips_to_move)

    # Ensure the destination folder exists
    os.makedirs(dest_folder, exist_ok=True)

    # Move the files
    for filename in selected:
        src = os.path.join(source_folder, filename)
        dst = os.path.join(dest_folder, filename)
        shutil.move(src, dst)
        print(f"Moved: {filename} → {dest_folder}")

# Regular expression to extract number of chips from folder name
chip_pattern = re.compile(r"(\d+)\s+chips", re.IGNORECASE)

# Iterate through each base directory
for region_key, region_base in base_paths.items():
    tif_folder = tif_sources[region_key]

    for folder in os.listdir(region_base):
        folder_path = os.path.join(region_base, folder)
        if os.path.isdir(folder_path) and "chips" in folder.lower():
            # Extract number of chips to move from folder name
            match = chip_pattern.search(folder)
            if not match:
                print(f"Skipping folder {folder}: could not determine number of chips to move.")
                continue

            num_chips = int(match.group(1))
            print(f"Processing folder: {folder} — moving {num_chips} chips.")

            # Move the chips
            move_chips(tif_folder, folder_path, num_chips)
```
